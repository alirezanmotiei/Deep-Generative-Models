# Energy-Based Models

```python
import torch
import torchvision
import torchvision.transforms as transforms
from torch.utils.data import DataLoader
import matplotlib.pyplot as plt
import numpy as np

transform = transforms.Compose([
    transforms.ToTensor(),
])


train_dataset = torchvision.datasets.MNIST(
    root='./data', 
    train=True, 
    download=True, 
    transform=transform
)

test_dataset = torchvision.datasets.MNIST(
    root='./data', 
    train=False, 
    download=True, 
    transform=transform
)


batch_size = 64

train_loader = DataLoader(
    dataset=train_dataset,
    batch_size=batch_size,
    shuffle=True,         
    num_workers=2,        
    pin_memory=True       
)

test_loader = DataLoader(
    dataset=test_dataset,
    batch_size=batch_size,
    shuffle=False,       
    num_workers=2,
    pin_memory=True
)

def show_batch(dataloader, title="Sample Images"):

    images, labels = next(iter(dataloader))
    
    num_images_to_show = 16
    images = images[:num_images_to_show]

    grid_img = torchvision.utils.make_grid(images, nrow=8, padding=2)

    np_img = grid_img.numpy().transpose((1, 2, 0))
    
    plt.figure(figsize=(12, 4))
    plt.imshow(np_img, cmap='gray')
    plt.title(title)
    plt.axis('off')
    plt.show()

print(f"Training Data Size: {len(train_dataset)}")
print(f"Test Data Size: {len(test_dataset)}")
print("-" * 40)

print("Displaying random images from Training set:")
show_batch(train_loader, title="Random Train Images (2x8 Grid)")

print("Displaying random images from Test set:")
show_batch(test_loader, title="Random Test Images (2x8 Grid)")
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader
import torchvision
import torchvision.transforms as transforms
import matplotlib.pyplot as plt
import numpy as np
import sys
import subprocess
from tqdm import tqdm

from torchinfo import summary

class EnergyModel(nn.Module):
    def __init__(self):
        super(EnergyModel, self).__init__()
        self.network = nn.Sequential(
            nn.Conv2d(1, 32, 3, 1, 1), nn.LeakyReLU(0.2),
            nn.Conv2d(32, 32, 4, 2, 1), nn.LeakyReLU(0.2),
            nn.Conv2d(32, 64, 3, 1, 1), nn.LeakyReLU(0.2),
            nn.Conv2d(64, 64, 4, 2, 1), nn.LeakyReLU(0.2),
            nn.Conv2d(64, 128, 3, 1, 1), nn.LeakyReLU(0.2),
            nn.AdaptiveAvgPool2d(1),
            nn.Flatten(),
            nn.Linear(128, 1)
        )

    def forward(self, x):
        return self.network(x)

def langevin_sampling(model, x, n_steps=60, step_size=10, show_pbar=False):
    iterator = range(n_steps)
    if show_pbar:
        iterator = tqdm(iterator, desc="Langevin Sampling", leave=False)

    for _ in iterator:
        with torch.enable_grad():
            x = x.detach()
            x.requires_grad = True
            energy = model(x)
            grad = torch.autograd.grad(energy.sum(), x, create_graph=False)[0]
        
        noise = torch.randn_like(x) * 0.005 
        x.data = x.data - step_size * grad + noise
        x.data.clamp_(0, 1)
        
    return x.detach()


def train_ebm():
    BATCH_SIZE = 64
    EPOCHS = 15           
    LAMBDA_REG = 1e-2       
    LR = 2 * 1e-3               
    DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

    transform = transforms.Compose([transforms.ToTensor()])
    train_dataset = torchvision.datasets.MNIST(root='./data', train=True, transform=transform, download=True)
    train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True, num_workers=2)

    model = EnergyModel().to(DEVICE)
    optimizer = optim.Adam(model.parameters(), lr=LR)


    scheduler = optim.lr_scheduler.ReduceLROnPlateau(
        optimizer, mode='min', factor=0.5, patience=3
    )

    print("\n" + "="*40)
    print("MODEL SUMMARY")
    print("="*40)
    dummy_input = torch.zeros((1, 1, 28, 28)).to(DEVICE)
    print(summary(model, input_data=dummy_input, col_names=["input_size", "output_size", "num_params"], verbose=0))
    print("="*40 + "\n")

    fixed_noise = torch.rand((16, 1, 28, 28)).to(DEVICE)

    print("Starting Training...")
    
    for epoch in range(EPOCHS):
        total_loss = 0.0
        total_e_real = 0.0
        total_e_fake = 0.0
        num_batches = 0
        
        pbar = tqdm(train_loader, desc=f"Epoch [{epoch+1}/{EPOCHS}]")
        
        for i, (x_real, _) in enumerate(pbar):
            x_real = x_real.to(DEVICE)
            curr_batch_size = x_real.shape[0]

            x_fake_init = torch.rand((curr_batch_size, 1, 28, 28)).to(DEVICE)
            x_fake = langevin_sampling(model, x_fake_init, n_steps=80, step_size=5, show_pbar=False)

            optimizer.zero_grad()
            e_real = model(x_real).squeeze()
            e_fake = model(x_fake).squeeze()

            data_term = e_real.mean() - e_fake.mean()
            reg_term = LAMBDA_REG * ( (e_real**2).mean() + (e_fake**2).mean() )
            loss = data_term + reg_term

            loss.backward()
            optimizer.step()
            
            total_loss += loss.item()
            total_e_real += e_real.mean().item()
            total_e_fake += e_fake.mean().item()
            num_batches += 1

            pbar.set_postfix(loss=f"{loss.item():.4f}")

        avg_loss = total_loss / num_batches
        avg_e_real = total_e_real / num_batches
        avg_e_fake = total_e_fake / num_batches
        
        scheduler.step(avg_loss)
        current_lr = optimizer.param_groups[0]['lr']

        print(f"Epoch [{epoch+1}/{EPOCHS}] Complete.")
        print(f"    Avg Loss:   {avg_loss:.4f}")
        print(f"    Avg E_Real: {avg_e_real:.4f}")
        print(f"    Avg E_Fake: {avg_e_fake:.4f}")
        print(f"    Current LR: {current_lr}")
        print("-" * 40)
        
        with torch.no_grad():
            samples = langevin_sampling(model, fixed_noise.clone(), n_steps=80, step_size=5, show_pbar=False)
            grid_img = torchvision.utils.make_grid(samples.cpu(), nrow=4, padding=2)
            plt.figure(figsize=(6,6))
            plt.imshow(grid_img.permute(1, 2, 0))
            plt.title(f"Generated Samples - Epoch {epoch+1}")
            plt.axis('off')
            plt.show()

    print("Final Sampling...")
    x_real_batch, _ = next(iter(train_loader))
    x_real_batch = x_real_batch[:8].to(DEVICE)
    
    grid_real = torchvision.utils.make_grid(x_real_batch.cpu(), nrow=8)
    plt.figure(figsize=(12, 2))
    plt.title("Original Real Images")
    plt.imshow(grid_real.permute(1, 2, 0))
    plt.axis('off')
    plt.show()
    
    final_samples = langevin_sampling(model, x_real_batch.clone(), n_steps=80, step_size=5, show_pbar=True)
    grid_fake = torchvision.utils.make_grid(final_samples.cpu(), nrow=8)
    plt.figure(figsize=(12, 2))
    plt.title("Langevin Samples (From Real)")
    plt.imshow(grid_fake.permute(1, 2, 0))
    plt.axis('off')
    plt.show()

    
    return model, train_loader




model, train_loader = train_ebm()
```

```python
import torch
import torchvision
import matplotlib.pyplot as plt

def generate_and_compare_samples(model, data_loader, device):

    model.eval()
    
    def plot_grid(tensor, title):
        grid_img = torchvision.utils.make_grid(tensor.cpu(), nrow=4, padding=2, normalize=True)
        plt.figure(figsize=(6, 6))
        plt.imshow(grid_img.permute(1, 2, 0), cmap='gray')
        plt.title(title)
        plt.axis('off')
        plt.show()

    print("1. Generating 16 samples starting from Uniform Noise...")
    noise = torch.rand(16, 1, 28, 28).to(device)
    fake_samples_from_noise = langevin_sampling(model, noise, n_steps=80, step_size=5)
    plot_grid(fake_samples_from_noise, "Generated (Start: Noise)")

    print("2. Displaying 16 Real Images from Train Data...")
    real_images, _ = next(iter(data_loader))
    real_images = real_images[:16].to(device)
    plot_grid(real_images, "Real Images (Start Points)")

    print("3. Generating 16 samples starting from Real Images...")
    fake_samples_from_real = langevin_sampling(model, real_images.clone(), n_steps=80, step_size=5)
    plot_grid(fake_samples_from_real, "Generated (Start: Real Data)")

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
generate_and_compare_samples(model, train_loader, device)
```

```python
import torch
import torchvision
import matplotlib.pyplot as plt

def run_denoising_experiment(model, test_loader, device):

    model.eval()
    
    x_original, _ = next(iter(test_loader))
    x_original = x_original[:16].to(device)
    
    noise_levels = [0.2, 0.4, 0.6]
    
    for noise_level in noise_levels:
        print(f"\n{'='*20} Noise Level: {noise_level} {'='*20}")
        

        noise = torch.randn_like(x_original) * noise_level
        x_noisy = x_original + noise
        
        x_noisy.clamp_(0, 1)
        

        x_denoised = langevin_sampling(model, x_noisy.clone(), n_steps=80, step_size=5)
        

        combined = torch.cat([x_original, x_noisy, x_denoised], dim=0)
        

        grid_img = torchvision.utils.make_grid(combined.cpu(), nrow=16, padding=2)
        
        plt.figure(figsize=(15, 5))
        plt.imshow(grid_img.permute(1, 2, 0), cmap='gray')
        plt.title(f"Top: Original | Middle: Noisy (Level {noise_level}) | Bottom: Denoised")
        plt.axis('off')
        plt.show()


device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
run_denoising_experiment(model, test_loader, device)
```
